#Label

```python
plt.plot(x,y)
plt.title('marksheet')
plt.xlabel('Name')
plt.ylabel('Marks')
plt.show()
```

we can work with font style

```python
font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}
plt.plot(x,y)
plt.title('marksheet',fontdict={'size':40}, loc='left')
plt.xlabel('Name', fontdict=font1)
plt.ylabel('Marks', fontdict=font2)
plt.show()
```

# Create the pie chart

```python
plt.pie(a)
plt.show()
```

add labels

```python
plt.pie(a,labels=a)
plt.show()
```

We can change the starting angle of this pai chart

```python
label=['apple', 'mango', 'banana','lichi', 'papaya', 'jackfruit']
plt.pie(a,labels=label,startangle=90)
plt.show()
```

```python
plt.pie(a,labels=label,startangle=40)
plt.show()
```

We can stand out an individual part

```python
plt.pie(a,labels=label,startangle=40,explode=[0.2,0,0,0,0,0])
plt.show()
```

```python
plt.pie(a,labels=label,startangle=40,explode=[0,0,0,0.5,0,0])
plt.show()
```

We can add shadow

```python
plt.pie(a,labels=label,startangle=40,explode=[0.5,0,0,0,0,0], shadow=True)
plt.show()
```

we can change the color manually

```python
color=['red','green','blue','yellow','pink','black']
plt.pie(a,labels=label,startangle=40,colors=color)
plt.show()
```

we can show all the label and percentage of this pie chart

```python
plt.pie(a,labels=label,startangle=40,colors=color,autopct='%1.1f%%')
plt.legend()
plt.show()
```

set titel

```python
plt.pie(a,labels=label,startangle=40,colors=color,autopct='%1.1f%%')
plt.legend(title='Fruits')
plt.show()
```
