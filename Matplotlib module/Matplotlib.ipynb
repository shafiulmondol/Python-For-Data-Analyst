#First import matplotlib.pyplot

```python
import matplotlib.pyplot as plt
```

This module just work on graph. We know in graph there are two axis x and y. Let us assume the value of x and y , and create our first graph with the value.

```python
x= [5,12,23,34,45,67,33]
y= [33,55,66,77,56,34,56]
```

For creating simple graph we use plt() function with peramiter of x and y. For showing this graph we use Show() function.

```python
plt.plot(x,y)
plt.show()
```

If we want to show just point we use one perameter in the plt function

```python
plt.plot(x,y,'o')
plt.show()
```
